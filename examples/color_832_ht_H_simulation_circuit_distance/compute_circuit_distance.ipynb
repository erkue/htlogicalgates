In this notebook we show that our circuits used for Fig. 4 have a circuit distance that is equal to two.

````python
import stim
from pysat.formula import WCNF
from pysat.examples.rc2 import RC2

def exact_circuit_distance(circuit: stim.Circuit) -> int:
    wdimacs = circuit.shortest_error_sat_problem()
    wcnf = WCNF(from_string=wdimacs)
    with RC2(wcnf) as rc2:
        model = rc2.compute()
        if model is None:
            raise RuntimeError("RC2 returned no model.")
        return rc2.cost
````

````python
print("Map 000 to +++ using H01, H1 with circuit distance:", exact_circuit_distance(stim.Circuit.from_file("000_to_ppp_H01H2.txt")))
print("Map +++ to 000 using H01, H1 with circuit distance:", exact_circuit_distance(stim.Circuit.from_file("ppp_to_000_H01H2.txt")))
print("Map 000 to +++ using H012    with circuit distance:", exact_circuit_distance(stim.Circuit.from_file("000_to_ppp_H012.txt")))
print("Map +++ to ppp using H012    with circuit distance:", exact_circuit_distance(stim.Circuit.from_file("ppp_to_000_H012.txt")))
````

Note that the provided circuits assume specific noise parameters, however, the output of the function ```shortest_error_sat_problem``` does not depend on these values.
